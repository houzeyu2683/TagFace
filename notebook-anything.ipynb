```python
import moviepy
import PIL.Image
import numpy
```

```python
path = 'rTsZQrNEP-Y_18_0.mkv'
video = moviepy.VideoFileClip(path)
```

```python
step = 10
delta = []
shot = []
for index, (timestep, frame) in enumerate(video.iter_frames(with_times=True)):
    if(timestep==0):
        history = frame
        shot += [frame]
        continue
    if(index%step!=0):
        continue
    delta += [frame - history]
    history = frame
    shot += [frame]
    continue
```

```python
PIL.Image.fromarray(shot[1])
```

```python
overview = numpy.concatenate([shot[0], delta[0], shot[0]+delta[0]], axis=1)
PIL.Image.fromarray(overview)
```
